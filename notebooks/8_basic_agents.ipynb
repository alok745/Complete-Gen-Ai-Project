```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
    It will return to be sum of two number
    
    Args:
        a: Number one
        b: Number two
    """
    return a+b



@tool
def multiply_numbers(a:int, b:int):
    """
    It will return to be multiply of two number
    
    Args:
        a: Number one
        b: Number two
    """
    return a+b
```

```python
add_numbers.invoke({"a":12,"b":5})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
agents = create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are a math teacher, and always use tool for calculation"
)
```

```python
res= agents.invoke({'messages':[{"role": "user","content": "What is address of kiet group of institutions  ?"}]})
print(res['messages'][-1].content)
```
