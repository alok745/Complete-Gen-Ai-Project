```python
from pydantic import BaseModel,Field
from typing import List, Annotated
```

```python
class User(BaseModel):
    name : str
    age : int
```

```python
user = User(name="Alok", age=24)
user
```

### Field validation 

```python
class Product(BaseModel):
    title :str = Field(max_length=10)
    des:str = Field(min_length=20)
    price:int= Field(gt=0, le=1000)
```

```python
product = Product(title="Samsung", des="This is amazing mobile  ", price=230)
product2= Product(title="Samsung", des="This is amazing mobile  ", price=230)
```

```python
class Order(BaseModel):
    id:str= Field(min_length=6, max_length=12)
    product:Product
```

```python
order = Order(id="dswds54542",product=product )
```

```python
order
```

```python
order.product.price
```

```python
class Cart(BaseModel):
    id:str= Field(min_length=6, max_length=12)
    items:List[Product]
    comment: str= Field(
        default="NO Comment"
    )
```

```python
cart = Cart(id="hhgsh1232", items=[product,product2],comment="this is my cart")
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

```python
### Annotated

commonValidator = Annotated[List[Product], "This is a Common Validator",Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str= Field(min_length=6, max_length=12)
    items:commonValidator
    comment: str= Field(
        default="NO Comment"
    )
```

```python
class Order(BaseModel):
    products: commonValidator
```

```python
ord = Order(products=[product,product2])
```
