```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model=" gemma4:31b-cloud",
    temperature=0,
)
```

```python

while True:
    query = input("user:")
    if query.lower() in ["exit", "bye", "quit"]:
     print("Good Bye")
     break
    res = llm.invoke(query)
    print(res.content)
```
