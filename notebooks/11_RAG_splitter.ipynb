### After loader split into chunks by 'Splitters'

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = "Today, I want to speak from the heart about three pillars that define the future of our nation, our youth, our education system, and our farmers."
```

```python
splitters = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap= 10)
```

```python
chunks = splitters.split_text(text)
chunks
```

### Document Loader

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loaders = TextLoader("../Data/speech.txt", encoding= "utf-8")
text = loaders.load()
len(text)
```

```python
pdfloader = PyPDFLoader("../Data/10 RAG Architectures.pdf")
pdfs = pdfloader.load()

pdfs
```

```python
splitters = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap= 200)
text_chunks = splitters.split_documents(text)
len(text_chunks)
```

### Text chunks

```python
splitters = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap= 100)
pdfs_chunks= splitters.split_documents(pdfs)
len(pdfs_chunks)
```
