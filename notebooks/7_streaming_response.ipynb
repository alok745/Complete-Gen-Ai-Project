```python
from langchain_groq import ChatGroq
```

### When We need response in chunk form we use 'Streaming' it give response like chatgpt 

```python
llm = ChatGroq(model="openai/gpt-oss-20b", streaming=True)
```

```python
query= "where is held brics 2026 ?'"
```

```python
res = llm.stream(query)
for chunk in res:
     print(chunk.content, end="")
```
