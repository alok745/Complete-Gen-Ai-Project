### Vector Embedding & vector Database

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    "Machine learning is a subset of Artificial intellegence",
    "Deep learning uses neural network with multiple layers",
    "Python is a programing language and it is useful for Data Science" ,
    "India is a beautiful country",
    "Ai models can understand and generate human language" 
]
```

```python
from langchain_ollama import OllamaEmbeddings

embedding = OllamaEmbeddings(
    model="nomic-embed-text:latest  "
)
```

```python
vector = embedding.embed_documents(documents)
vector 
```

### Store Vector Embedding in Vector DB

```python
from langchain_chroma import Chroma
vector_store = Chroma.from_texts(
    texts= documents,
    embedding=embedding,
    persist_directory="./vector.db"
    
)
```

```python
query = "Country"
result = vector_store.similarity_search(query=query, k=2)
result
```
